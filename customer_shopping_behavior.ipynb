```python
df.columns=df.columns.str.lower()
```

```python
df.columns=df.columns.str.replace('purchase_amount_(usd)','purchase_amount')
```

```python
df.isnull().sum()
```

```python
df['review_rating']=df.groupby('category')['review_rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#age group column 
labels=['young adult','adult','middle-aged','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df['frequency_of_purchases'].unique()
```

```python
#purches frequency days column creation 
frequency_mapping={'Fortnightly':14,'Weekly':7, 'Annually':365, 'Quarterly':90, 'Bi-Weekly':14,
       'Monthly':30, 'Every 3 Months' :90}
df['purches_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)    
```

```python
from sqlalchemy import create_engine, text

engine = create_engine(
    "mysql+pymysql://root:123@localhost:3306/customer_behavior"
)


# 🔹 Drop old table (important to avoid mismatch)
with engine.connect() as conn:
    conn.execute(text("DROP TABLE IF EXISTS customer"))

# 🔹 Create table dynamically from your df
columns = ", ".join([f"`{col}` TEXT" for col in df.columns])

with engine.connect() as conn:
    conn.execute(text(f"CREATE TABLE customer ({columns})"))

# 🔹 Insert all data
data = df.to_dict(orient="records")

with engine.begin() as conn:
    conn.execute(
        text(f"""
            INSERT INTO customer ({', '.join([f"`{col}`" for col in df.columns])})
            VALUES ({', '.join([f":{col}" for col in df.columns])})
        """),
        data
    )

print("✅ Original data inserted successfully")
```
